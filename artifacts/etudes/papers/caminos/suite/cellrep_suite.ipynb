## Built with jaxfne

The Cell Reports network written in the jaxfne grammar: `Configuration` → `construct` → `Model.simulate`, with
the per-type AdEx values of the Cell Reports set (`caminos_api.py`). This is the core route,
`run_cell(route="core")`. The figures below come from the oracle runs (`caminos_jax.py`, the authors' model
transcribed to JAX). The core route is not yet equivalent to them: E1 at n 12 found 20 of 24 cells equivalent
within ±5% (`RESULT_FIGURES.md`). These cells show jaxfne usage; they are not the source of the figures. They run
1 s of network time live.

```python
sys.path.insert(0, str(Path(cs.__file__).resolve().parents[1]))   # caminos/
import caminos_api as ca      # the Cell Reports constants; puts the checkout's jaxfne first
import jaxfne as jtfne

p, seed = ca.CELLREP_P, 1     # p 0.017 (E -> PV at twice p), seed
n_e, n_pv, n_som = ca.CELLREP_NE, ca.CELLREP_NPV, ca.CELLREP_NSOM
n = n_e + n_pv + n_som
```

Populations (8000 E, 800 PV, 600 SOM), the AdEx emitter, one AMPA receptor per target type, GABA, and the per-type AdEx values.

```python
cfg = (jtfne.Configuration()
       .network(name="cellrep", n=n, cell_types={"E": n_e / n, "PV": n_pv / n, "SOM": n_som / n})
       .emitter(family="adex", adex_decay="euler", adex_step_order="default"))
for name, tau_ms in (("ampa_e", ca.CELLREP_TAU_E_E_MS), ("ampa_pv", ca.CELLREP_TAU_E_PV_MS),
                     ("ampa_som", ca.CELLREP_TAU_E_SOM_MS)):
    cfg = cfg.mechanisms(name=name, kind="receptor", params={"tau_ms": tau_ms, "reversal_mV": ca.CELLREP_REV_E_MV})
cfg = cfg.mechanisms(name="gaba", kind="receptor", params={"tau_ms": ca.CELLREP_TAU_I_MS, "reversal_mV": ca.CELLREP_REV_I_MV})
cfg = (cfg.cell_params({"cell_type": "E"}, {"EL_mV": -60.0, "b_pA": 130.0})
          .cell_params({"cell_type": "PV"}, {"EL_mV": -60.0})
          .cell_params({"cell_type": "SOM"}, {"Delta_mV": 1.5, "b_pA": 25.0}))
```

Seven Bernoulli connection rules (no PV→SOM, no SOM→SOM; E→PV at twice p) with a one-step delay, then the
check that this network equals the core route's.

```python
for name, pre, post, weight_nS, receptor, mult in ca.CELLREP_RULES:
    cfg = cfg.connections(name=name, source={"cell_type": pre}, target={"cell_type": post},
                          probability=p * mult, weight=weight_nS, mechanism=receptor,
                          delay_ms=ca.DT_MS, sampling="bernoulli")
cfg = cfg.connectivity(edge_seed=seed)
assert cfg == ca.build_cellrep_spec(p=p, seed=seed), "the network must equal the core route's"
```

Construct the model and simulate 1 s with the Poisson drive (136 sources at 4 Hz onto E and PV).

```python
model = jtfne.construct(cfg)
drive = ca.cellrep_drive_entries(p, n_e, seed)
signals = model.simulate(jtfne.Simulation(duration_ms=1000.0, dt_ms=ca.DT_MS, poisson_drive=drive))
_ = cs.core_run_figure(signals.spikes, (n_e, n_pv, n_som),
                       "jaxfne core run: seed 1 (last 0.5 s of 1 s)", source="live run in this notebook")
```
