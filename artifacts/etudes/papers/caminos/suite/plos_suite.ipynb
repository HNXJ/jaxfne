## Built with jaxfne

The CAMINOS network of the PLoS paper written in the jaxfne grammar: `Configuration` → `construct` →
`Model.simulate`. The values come from the paper's Tables 1-2 (`caminos_api.py`, `PORT_SPEC.md`). This network
is the P6 replication of Fig 1, which passed through the jaxfne public API (`RESULT_P6.md`); Fig 1E below reads
those P6 runs. The other panels read the oracle runs (`caminos_jax.py`, the authors' model transcribed to JAX)
or the mean-field model. The cells below run 1 s of network time live.

```python
import jaxfne as jtfne
import caminos_api as ca      # the paper's constants (Tables 1-2)
import cellrep_suite as cs    # raster plotting

p, seed = 0.04, 1             # connection probability (Fig 1 example), seed
```

Populations (8000 E, 750 PV, 750 SOM), the AdEx emitter and the two receptors.

```python
cfg = (jtfne.Configuration()
       .network(name="caminos", n=ca.N, cell_types={"E": ca.NE / ca.N, "PV": ca.NPV / ca.N, "SOM": ca.NSOM / ca.N})
       .emitter(family="adex", adex_decay="euler", adex_step_order="default")
       .mechanisms(name="ampa", kind="receptor", params={"tau_ms": ca.TAU_E_MS, "reversal_mV": ca.REV_E_MV})
       .mechanisms(name="gaba", kind="receptor", params={"tau_ms": ca.TAU_I_MS, "reversal_mV": ca.REV_I_MV}))
```

Seven Bernoulli connection rules (E→E, E→PV, E→SOM, PV→E, PV→PV, SOM→E, SOM→PV) with the 2 ms delay, then
the check that this network equals the one the P6 runs used.

```python
for name, pre, post, weight_nS, receptor in ca.RULES:
    cfg = cfg.connections(name=name, source={"cell_type": pre}, target={"cell_type": post},
                          probability=p, weight=weight_nS, mechanism=receptor,
                          delay_ms=ca.DELAY_MS, sampling="bernoulli")
cfg = cfg.connectivity(edge_seed=seed)
assert cfg == ca.build_spec(p, seed), "the network must equal the one the P6 runs used"
```

Construct the model and simulate 1 s with the Poisson drive (p·NE sources at 4 Hz onto E and PV).

```python
model = jtfne.construct(cfg)
sim = jtfne.Simulation(duration_ms=1000.0, dt_ms=ca.DT_MS, poisson_drive=ca.drive_spec(p, seed))
signals = model.simulate(sim)
fig = cs.core_run_figure(signals.spikes, (ca.NE, ca.NPV, ca.NSOM),
                         "jaxfne core run: p 0.04, seed 1 (last 0.5 s of 1 s)", source="live run in this notebook")
display(fig)
plt.close(fig)
```
